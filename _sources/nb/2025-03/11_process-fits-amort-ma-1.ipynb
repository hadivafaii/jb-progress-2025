````python
root = 'Dropbox/chkpts/PoissonVAE'
root = add_home(root)

fits = sorted(os.listdir(root), key=alphanum_sort_key)
````

````python
print(fits)
````

````python
# the big for loop
for name in tqdm(fits):
    save_name = f"{name}.npy"
    if os.path.isfile(pjoin(results_dir, save_name)):
        continue

    # (1) load trainer
    try:
        tr, meta = load_model_lite(
            pjoin(root, name),
            device=device,
            shuffle=False,
            verbose=False,
            strict=True,
        )
    except StopIteration:
        print(f"missing: {name}")
        continue
    
    break
````

````python
# compute loss, xtract ftrs
data, loss, etc = tr.validate(full_data=True)
loss['mse_map'] = tr.mse_map()
````
