````python
# HIDE CODE


def is_fit_on_or_after(fit_name, threshold_date):
    if threshold_date is None:
        return True

    if '(' in fit_name and ')' in fit_name:
        date_part = fit_name.split('(')[-1].split(')')[0]
        fit_date = date_part.split(',')[0]
    else:
        for part in fit_name.split('_'):
            if part.isdigit() and len(part) == 4:
                year_index = fit_name.find(part)
                if year_index >= 0 and len(fit_name) >= year_index + 10:
                    fit_date = fit_name[year_index:year_index+10]
                    if fit_date.count('_') == 2:
                        break
        else:
            return False
    
    fit_date_compact = fit_date.replace('_', '')
    threshold_date_compact = threshold_date.replace('_', '')
    
    return fit_date_compact >= threshold_date_compact


def get_fit_status(dataset: str, threshold_date: str = None):
    status = {}
    
    for model_name in model_names:
        if dataset not in model_name:
            continue
        path = pjoin(fits_path, model_name)
        fit_names = [
            d for d in os.listdir(path) if
            os.path.isdir(pjoin(path, d))
        ]
        fit_names = sorted(fit_names, key=alphanum_sort_key)
        for fit_name in fit_names:
            if not is_fit_on_or_after(fit_name, threshold_date):
                continue
            path = pjoin(fits_path, model_name, fit_name)
            checkpoints = [f for f in os.listdir(path) if f.endswith('.pt')]
            checkpoints = sorted(checkpoints, key=alphanum_sort_key)
            try:
                last_checkpoint = checkpoints[-1]
                last_checkpoint = int(last_checkpoint.split('_')[0].split('-')[-1])
            except IndexError:
                last_checkpoint = None
    
            status[(model_name, fit_name)] = last_checkpoint

    return status
````
