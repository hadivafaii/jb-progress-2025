````python
device_idx = 1
device = f'cuda:{device_idx}'

print(f"device: {device}  ———  host: {os.uname().nodename}")
````

## Results dir

````python
results_dir = pjoin(tmp_dir, 'results_2025-03')
os.makedirs(results_dir, exist_ok=True)

len(os.listdir(results_dir))
````

## Fits

````python
root = '/mnt/sata/Dekel/vH16/LCA/runs'

fits = {}
for f in os.listdir(root):
    lamb = f.split('_')[1].split('-')[1].split(':')[1]
    fits[lamb] = f
fits = dict(sorted(
    fits.items(),
    key=lambda tup: float(tup[0]),
))
len(fits)
````

````python
vld = '/mnt/sata/Dekel/vH16/vld.npy'
vld = torch.tensor(np.load(vld), device=device)
vld = vld.flatten(start_dim=1)
vld.shape
````

## Process

````python
for lamb, f in fits.items():
    run = dict(np.load(pjoin(root, f)))

    # display the fit
    print(f"\n\n run name:\n{f}")
    w = run['dictionary'].T
    w = w.reshape(-1, 16, 16)
    plot_weights(w)

    # info
    info = {
        'str_model': 'lca',
        'seq_len': -1,
        'kl_beta': float(lamb),
    }
    # results
    r2 = compute_r2(
        true=vld,
        pred=torch.tensor(run['recons'], device=device),
    ).mean().item()
    results = {'r2': r2, 'r2_map_final': r2}
    results['%-zeros'] = (run['activations'] == 0).mean()

    print({'info': info, 'results': results})

    save_obj(
        obj={'info': info, 'results': results},
        file_name=f.replace('.npz', '.npy'),
        save_dir=results_dir,
        verbose=True,
    )
    print('-' * 90)
````
