````python
model_type = 'poisson'
cfg_vae, cfg_tr = default_configs('CelebA', model_type, 'ngd|deep')

t_train = 8

cfg_vae['t_train'] = t_train
cfg_vae['kl_beta'] = 1/8 * t_train

cfg_vae['latent_channels'] = 256
cfg_vae['latent_pixels'] = 16

cfg_tr['batch_size'] = 40
cfg_tr['epochs'] = 30

cfg_tr['eval_freq'] = 1
cfg_tr['chkpt_freq'] = 1
````
